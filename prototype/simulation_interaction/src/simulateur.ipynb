```python
import turtle as tl
import time
```

```python
tl.clearscreen()

tl.up()
tl.setheading(90)
tl.speed(3)

while(True):    
    orders: list[str] = []
    with(open("channel.txt", "r+") as channel):
        for line in channel:
            print(line)
            orders.append(line)
        channel.truncate(0)

    for order in orders:
        if(order.startswith("FORWARD")):
            tl.forward(50)
        elif(order.startswith("BACKWARD")):
            tl.backward(50)
        elif(order.startswith("LEFT")):
            tl.setheading(tl.heading() + 90)
        elif(order.startswith("RIGHT")):
            tl.setheading(tl.heading() - 90)

    time.sleep(0.5)
```

```python
# Close the turtle window
tl.bye()
```
